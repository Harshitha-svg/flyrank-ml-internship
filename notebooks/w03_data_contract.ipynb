# W03 Data Contract


## 1. Data Contract
One row represents a page-date observation.


```
SELECT page,date,COUNT(*) AS duplicate_count FROM search_console_daily WHERE month='2026-03' GROUP BY page,date HAVING COUNT(*)>1;
```

```
SELECT COUNT(*) AS row_count, MIN(date) AS start_date, MAX(date) AS end_date FROM search_console_daily WHERE month='2026-03';
```

```
SELECT COUNT(*) AS available_rows FROM search_console_daily WHERE month='2026-03' AND is_available IS TRUE;
```
