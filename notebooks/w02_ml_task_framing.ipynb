# ML-03: Frame Your Lane as an ML Task


## 1. My Lane as an ML Task

**Task Type:** Ranking

The goal is to rank content based on relevance and quality so the best content can be surfaced to users.

## 2. Target or Proxy

**Target/Proxy:** Content relevance score.

A higher score indicates content that is more likely to satisfy user intent and perform well.

## 3. Success Metric

**Metric:** NDCG (Normalized Discounted Cumulative Gain).

This measures how well the ranking places the most relevant content at the top.

````
import pandas as pd

df = pd.DataFrame({
    "content_id":[1,2,3,4],
    "title":["ML Basics","SEO Guide","Python Tips","Data Science Intro"],
    "engagement_score":[82,74,91,68]
})

df.head()
````

## 4. Unit of Analysis

One row = one content item/article.

Example target column:

```python
df["target_score"] = df["engagement_score"]
```

## 5. Why ML Beats a Fixed Rule

A fixed rule cannot capture complex interactions between content quality, relevance, engagement, and user intent. ML can learn patterns from multiple signals and adapt as data changes.

## 6. Real Content Action

The ranking output helps decide which content should be promoted, optimized, or recommended to users.
